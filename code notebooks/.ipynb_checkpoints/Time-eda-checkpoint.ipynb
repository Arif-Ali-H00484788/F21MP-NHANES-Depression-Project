```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from pygam import LogisticGAM, s
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, accuracy_score, f1_score
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
import pygam as pg
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.multiclass import OneVsRestClassifier
from pygam import LogisticGAM, s
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.utils import to_categorical
import shap
import warnings
from functools import reduce
import operator

# Wrapping pygam’s LogisticGAM in a scikit‑learn API
class GAMClassifier(BaseEstimator, ClassifierMixin):
    def __init__(self, terms, **gam_kwargs):
        self.terms = terms
        self.gam_kwargs = gam_kwargs

    def fit(self, X, y):
        # fit a binary LogisticGAM
        self.gam_ = LogisticGAM(self.terms, **self.gam_kwargs).fit(X, y)
        return self

    def predict(self, X):
        return self.gam_.predict(X)

    def predict_proba(self, X):
        # pygam.predict_proba returns p(class=1) as a 1D array
        p1 = self.gam_.predict_proba(X)
        # stack to get shape (n_samples, 2): [p(class=0), p(class=1)]
        return np.vstack([1 - p1, p1]).T

warnings.filterwarnings('ignore')
sns.set_style('whitegrid')

# ## 2. Data Loading and Exploration
#
# We will load the dataset and perform an initial exploration to understand its structure, features, and the nature of the longitudinal data.

# Load the dataset from the uploaded CSV file
try:
    PROC_URL = ("https://raw.githubusercontent.com/Arif-Ali-H00484788/F21MP-NHANES-Depression-Project/main/data/processed/Model%20Training-NHANES%20Depression%20Screening%20Dataset.csv")
    df = pd.read_csv(PROC_URL)
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Error: 'Model Training-NHANES Depression Screening Dataset.csv' not found.")
    print("Please ensure the dataset file is in the same directory as this notebook.")
    # As a fallback for demonstration, create a dummy dataframe
    df = pd.DataFrame() # Or exit

if not df.empty:
    print("\nDataset Information:")
    df.info()

    print("\nFirst 5 Rows of the Dataset:")
    print(df.head())

    print("\nDescriptive Statistics:")
    print(df.describe())

    # --- Visualize Patient Trajectories ---
    plt.figure(figsize=(14, 7))
    # Get a sample of 10 patients to visualize their PHQ-9 score trajectories
    sample_patient_ids = df['Patient_Identifier'].unique()[:10]
    for patient_id in sample_patient_ids:
        patient_data = df[df['Patient_Identifier'] == patient_id]
        plt.plot(patient_data['Week'], patient_data['PHQ9_Score'], marker='o', linestyle='-', label=f'Patient {patient_id}')
    plt.title('PHQ-9 Score Trajectory for Sample Patients (12 Weeks)')
    plt.xlabel('Week')
    plt.ylabel('PHQ-9 Score')
    plt.xticks(range(1, 13))
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()


# ## 3. Data Preprocessing and Feature Engineering
#
# This is a critical step. We need to prepare the data for both traditional ML models and sequence-based models like LSTMs.
#
# **Strategy:**
# 1.  **Encode Targets:** Convert categorical target variables (Depression_Severity, Final_Risk) to numerical labels.
# 2.  **Patient-Wise Splitting:** Use GroupShuffleSplit to ensure that all 12 records for any given patient belong to *either* the training set or the testing set. This prevents data leakage and respects the longitudinal structure.
# 3.  **Prepare Data for LSTM:** Reshape the data into a 3D tensor of shape (n_patients, n_timesteps, n_features).

# --- Target Variable Encoding ---
# Initialize encoders for our two target variables
sev_encoder = LabelEncoder()
risk_encoder = LabelEncoder()

# Since each patient has the same severity/risk over 12 weeks, we can encode based on the first week's value
# and then apply it to all weeks for that patient.
df['Depression_Severity_Encoded'] = sev_encoder.fit_transform(df.groupby('Patient_Identifier')['Depression_Severity'].transform('first'))
df['Final_Risk_Encoded'] = risk_encoder.fit_transform(df.groupby('Patient_Identifier')['Final_Risk'].transform('first'))

# Store mappings for later interpretation
sev_classes = sev_encoder.classes_
risk_classes = risk_encoder.classes_
print(f"Depression Severity Classes: {dict(enumerate(sev_classes))}")
print(f"Final Risk Classes: {dict(enumerate(risk_classes))}")

# --- Feature Selection ---
# For our models, we will use the individual questionnaire answers and the total score.
features = [f'DPQ{i}' for i in range(1, 11)] + ['PHQ9_Score']
X = df[features]
y_sev = df['Depression_Severity_Encoded']
y_risk = df['Final_Risk_Encoded']
groups = df['Patient_Identifier'] # This is crucial for the split

# --- Patient-Wise Train-Test Split ---
# We use GroupShuffleSplit to keep all data from a single patient in the same set
splitter = GroupShuffleSplit(test_size=0.2, n_splits=1, random_state=42)
train_idx, test_idx = next(splitter.split(X, y_sev, groups))

# Traditional ML Data
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_sev_train, y_sev_test = y_sev.iloc[train_idx], y_sev.iloc[test_idx]
y_risk_train, y_risk_test = y_risk.iloc[train_idx], y_risk.iloc[test_idx]

print(f"\nData Split:")
print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
print(f"Number of patients in training set: {df.iloc[train_idx]['Patient_Identifier'].nunique()}")
print(f"Number of patients in testing set: {df.iloc[test_idx]['Patient_Identifier'].nunique()}")

# --- LSTM Data Preparation ---
# LSTMs require data in a 3D tensor format: (samples, timesteps, features)
# Samples = number of patients
# Timesteps = 12 (weeks)
# Features = number of features used
def create_sequence_data(df, cols, target, group, idx):
    Xs, ys = [], []
    for _, g in df.iloc[idx].groupby(group):
        g = g.sort_values('Week')
        Xs.append(g[cols].values)
        ys.append(g[target].iloc[0])
    return np.array(Xs), np.array(ys)

X_sev_train_seq, y_sev_train_seq = create_sequence_data(
    df, features, 'Depression_Severity_Encoded', 'Patient_Identifier', train_idx)
X_sev_test_seq, y_sev_test_seq = create_sequence_data(
    df, features, 'Depression_Severity_Encoded', 'Patient_Identifier', test_idx)
X_risk_train_seq, y_risk_train_seq = create_sequence_data(
    df, features, 'Final_Risk_Encoded', 'Patient_Identifier', train_idx)
X_risk_test_seq, y_risk_test_seq = create_sequence_data(
    df, features, 'Final_Risk_Encoded', 'Patient_Identifier', test_idx)

print(f"\nLSTM sequence data shape for Severity (Train): {X_sev_train_seq.shape}")
print(f"LSTM target data shape for Severity (Train): {y_sev_train_seq.shape}")


# ## 4. Model Training and Evaluation
#
# We will now train and evaluate each model for both target variables. We'll store the results in a dictionary for easy comparison later.
#
# ### 4.1 Target 1: Depression Severity

# A helper function to plot confusion matrices
def plot_confusion_matrix(y_true, y_pred, classes, title):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
    plt.title(title)
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.show()

results = {} # To store model performance

# 1. Scale features so DPQ1–DPQ10 (0–3) and PHQ9_Score (0–27) are on the same scale
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

# 2. Define spline terms once
terms = reduce(operator.add, (s(i) for i in range(X_train_scaled.shape[1])))


# 3. Hyperparameter grids
lam_grid       = np.logspace(-2, 2, 5)   # [0.01, 0.1, 1, 10, 100]
n_splines_grid = [8, 16, 24]            # try small, medium, large bases

# 4. Function to train one‐vs‐rest binary GAMs and return dict[label] = fitted GAM
def train_multiclass_gams(X, y):
    unique_labels = np.unique(y)
    gams = {}
    for lbl in unique_labels:
        # binary target for this label
        y_bin = (y == lbl).astype(int)
        # initialize and grid‐search
        gam = LogisticGAM(terms, fit_intercept=True)
        gam = gam.gridsearch(X, y_bin, lam=lam_grid, n_splines=n_splines_grid)
        gams[lbl] = gam
    return unique_labels, gams


# --- Model 1: Generalized Additive Model (GAM) ---

print("\n--- Training GAM for Depression Severity   ---")
sev_labels, sev_gams = train_multiclass_gams(X_train_scaled, y_sev_train)

# stack p(class=1) for each binary GAM, then argmax to get multiclass pred
sev_proba = np.vstack([gam.predict_proba(X_test_scaled) for gam in sev_gams.values()]).T
gam_sev_preds = sev_labels[np.argmax(sev_proba, axis=1)]

print(classification_report(y_sev_test, gam_sev_preds, target_names=sev_classes))
results['GAM (Severity)'] = {
    'Accuracy': accuracy_score(y_sev_test, gam_sev_preds),
    'F1-Score': f1_score(y_sev_test, gam_sev_preds, average='weighted')
}
plot_confusion_matrix(y_sev_test, gam_sev_preds, sev_classes, 'GAM (OvR) Confusion Matrix (Severity)')

# --- Model 2: Random Forest ---
print("\n--- Training Random Forest for Depression Severity ---")
rf_sev = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_sev.fit(X_train, y_sev_train)
rf_sev_preds = rf_sev.predict(X_test)

print(classification_report(y_sev_test, rf_sev_preds, target_names=sev_classes))
results['Random Forest (Severity)'] = {
    'Accuracy': accuracy_score(y_sev_test, rf_sev_preds),
    'F1-Score': f1_score(y_sev_test, rf_sev_preds, average='weighted')
}
plot_confusion_matrix(y_sev_test, rf_sev_preds, sev_classes, 'Random Forest Confusion Matrix (Severity)')


# --- Model 3: XGBoost ---
print("\n--- Training XGBoost for Depression Severity ---")
xgb_sev = xgb.XGBClassifier(objective='multi:softmax', num_class=len(sev_classes), eval_metric='mlogloss', use_label_encoder=False, random_state=42)
xgb_sev.fit(X_train, y_sev_train)
xgb_sev_preds = xgb_sev.predict(X_test)

print(classification_report(y_sev_test, xgb_sev_preds, target_names=sev_classes))
results['XGBoost (Severity)'] = {
    'Accuracy': accuracy_score(y_sev_test, xgb_sev_preds),
    'F1-Score': f1_score(y_sev_test, xgb_sev_preds, average='weighted')
}
plot_confusion_matrix(y_sev_test, xgb_sev_preds, sev_classes, 'XGBoost Confusion Matrix (Severity)')


# --- Model 4: LSTM ---
print("\n--- Training LSTM for Depression Severity ---")
y_sev_train_cat = to_categorical(y_sev_train_seq, num_classes=len(sev_classes))
y_sev_test_cat = to_categorical(y_sev_test_seq, num_classes=len(sev_classes))

lstm_sev = Sequential([
    LSTM(64, input_shape=(X_sev_train_seq.shape[1], X_sev_train_seq.shape[2]), return_sequences=True),
    Dropout(0.3),
    LSTM(32),
    Dropout(0.3),
    Dense(len(sev_classes), activation='softmax')
])
lstm_sev.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
history = lstm_sev.fit(X_sev_train_seq, y_sev_train_cat, epochs=25, batch_size=16, validation_split=0.1, verbose=0)
lstm_sev_pred_probs = lstm_sev.predict(X_sev_test_seq)
lstm_sev_preds = np.argmax(lstm_sev_pred_probs, axis=1)

print(classification_report(y_sev_test_seq, lstm_sev_preds, target_names=sev_classes))
results['LSTM (Severity)'] = {
    'Accuracy': accuracy_score(y_sev_test_seq, lstm_sev_preds),
    'F1-Score': f1_score(y_sev_test_seq, lstm_sev_preds, average='weighted')
}
plot_confusion_matrix(y_sev_test_seq, lstm_sev_preds, sev_classes, 'LSTM Confusion Matrix (Severity)')


# ### 4.2 Target 2: Final Risk
#
# Now we repeat the process for the Final_Risk target, applying the same correction.

print("\n" + "="*50)
print("### Training Models for Final Risk ###")
print("="*50 + "\n")

# --- Model 1: GAM ---
print("\n--- Training GAM (OvR) for Final Risk ---")
risk_labels, risk_gams = train_multiclass_gams(X_train_scaled, y_risk_train)

risk_proba = np.vstack([gam.predict_proba(X_test_scaled) for gam in risk_gams.values()]).T
gam_risk_preds = risk_labels[np.argmax(risk_proba, axis=1)]

print(classification_report(y_risk_test, gam_risk_preds, target_names=risk_classes))
results['GAM (Risk)'] = {
    'Accuracy': accuracy_score(y_risk_test, gam_risk_preds),
    'F1-Score': f1_score(y_risk_test, gam_risk_preds, average='weighted')
}
plot_confusion_matrix(y_risk_test, gam_risk_preds, risk_classes, 'GAM (OvR) Confusion Matrix (Risk)')

# --- Model 2: Random Forest ---
print("\n--- Training Random Forest for Final Risk ---")
rf_risk = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1).fit(X_train, y_risk_train)
rf_risk_preds = rf_risk.predict(X_test)
print(classification_report(y_risk_test, rf_risk_preds, target_names=risk_classes))
results['Random Forest (Risk)'] = {'Accuracy': accuracy_score(y_risk_test, rf_risk_preds), 'F1-Score': f1_score(y_risk_test, rf_risk_preds, average='weighted')}

# --- Model 3: XGBoost ---
print("\n--- Training XGBoost for Final Risk ---")
xgb_risk = xgb.XGBClassifier(objective='multi:softmax', num_class=len(risk_classes), eval_metric='mlogloss', use_label_encoder=False, random_state=42).fit(X_train, y_risk_train)
xgb_risk_preds = xgb_risk.predict(X_test)
print(classification_report(y_risk_test, xgb_risk_preds, target_names=risk_classes))
results['XGBoost (Risk)'] = {'Accuracy': accuracy_score(y_risk_test, xgb_risk_preds), 'F1-Score': f1_score(y_risk_test, xgb_risk_preds, average='weighted')}

# --- Model 4: LSTM ---
print("\n--- Training LSTM for Final Risk ---")
y_risk_train_cat = to_categorical(y_risk_train_seq, num_classes=len(risk_classes))
y_risk_test_cat = to_categorical(y_risk_test_seq, num_classes=len(risk_classes))
lstm_risk = Sequential([
    LSTM(64, input_shape=(X_risk_train_seq.shape[1], X_risk_train_seq.shape[2]), return_sequences=True), Dropout(0.3),
    LSTM(32), Dropout(0.3),
    Dense(len(risk_classes), activation='softmax')
])
lstm_risk.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
lstm_risk.fit(X_risk_train_seq, y_risk_train_cat, epochs=25, batch_size=16, validation_split=0.1, verbose=0)
lstm_risk_pred_probs = lstm_risk.predict(X_risk_test_seq)
lstm_risk_preds = np.argmax(lstm_risk_pred_probs, axis=1)
print(classification_report(y_risk_test_seq, lstm_risk_preds, target_names=risk_classes))
results['LSTM (Risk)'] = {'Accuracy': accuracy_score(y_risk_test_seq, lstm_risk_preds), 'F1-Score': f1_score(y_risk_test_seq, lstm_risk_preds, average='weighted')}


# ## 5. Performance Comparison
#
# Let's compile all the results into a single table and visualize them to compare the models' performance head-to-head.

results_df = pd.DataFrame(results).T.reset_index()
results_df = results_df.rename(columns={'index': 'Model'})
results_df[['Model', 'Target']] = results_df['Model'].str.extract(r'(.+?) \((.+?)\)')

print("\n--- Overall Model Performance Comparison ---")
print(results_df.to_markdown(index=False))

# Visualize the comparison
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)
sns.barplot(x='Model', y='Accuracy', hue='Target', data=results_df, ax=axes[0])
axes[0].set_title('Model Comparison by Accuracy')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=45, ha='right')

sns.barplot(x='Model', y='F1-Score', hue='Target', data=results_df, ax=axes[1])
axes[1].set_title('Model Comparison by Weighted F1-Score')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45, ha='right')

plt.tight_layout()
plt.show()


# ## 6. Model Interpretability
#
# Performance is important, but for clinical settings, interpretability is key. We need to understand *why* a model makes a certain prediction.
#
# * **GAMs** are inherently interpretable.
# * For **Tree-based models** and **LSTMs**, we use **SHAP** (SHapley Additive exPlanations).

# ### 6.1 GAM Interpretability
#
# GAMs allow us to plot the relationship (spline) learned for each feature.
print("\n--- Interpreting the GAM for Final Risk (class‑specific splines) ---")

n_classes = len(risk_classes)
n_feats   = len(features)

# create a grid: one row per class, one col per feature
fig, axs = plt.subplots(n_classes, n_feats, figsize=(4 * n_feats, 3 * n_classes), squeeze=False)

for k, lbl in enumerate(risk_labels):
    # lbl is the encoded class label (0, 1, 2, …)
    gam_k = risk_gams[lbl]                       # <-- use your fitted dict
    class_name = risk_classes[lbl]               # map back to the string name
    for i, feat in enumerate(features):
        ax = axs[k, i]
        XX = gam_k.generate_X_grid(term=i)
        ax.plot(XX[:, i], gam_k.partial_dependence(term=i, X=XX))
        ax.set_title(f"{class_name}\n{feat}")
        if k == n_classes - 1:
            ax.set_xlabel(feat)
        if i == 0:
            ax.set_ylabel("Effect")

fig.suptitle('GAM Partial Dependence by Class (Final Risk)', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

# ### 6.2 SHAP for Black-Box Models
#
# We will focus on the best-performing "black-box" model, likely XGBoost, for the Final_Risk target.
print("\n--- Interpreting XGBoost with SHAP ---")
# Create a SHAP explainer for the XGBoost model
explainer = shap.TreeExplainer(xgb_risk)
shap_values = explainer.shap_values(X_test)

# SHAP Summary Plot: Shows global feature importance
shap.summary_plot(shap_values, X_test, plot_type="bar", class_names=risk_classes, feature_names=features)
shap.summary_plot(shap_values, X_test, class_inds='original', feature_names=features)

# SHAP Force Plot: Explains a single prediction
# We'll explain the first prediction in the test set
print("\nExplaining a single prediction with a SHAP Force Plot:")

# pick which class and which instance to explain
class_idx    = 0     # e.g. 0 for the first class
instance_idx = 0     # first sample in the test set

# extract the SHAP vector for that class & instance
sv         = shap_values[class_idx][instance_idx]        # numpy array, shape = (n_features,)
base_val   = explainer.expected_value[class_idx]         # scalar
feat_vals  = X_test.iloc[instance_idx].to_numpy()        # numpy array, shape = (n_features,)

# sanity check
print("shap vector length:", sv.shape[0])
print("feature array length:", feat_vals.shape[0])
print("feature_names length:", len(features))

# now all three should match:
#   shap vector length == feature array length == len(features)

# draw the force plot
shap.force_plot(
    base_val,
    sv,
    feat_vals,
    feature_names=features,
    matplotlib=True
)
```
